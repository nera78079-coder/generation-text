```python
import tensorflow as tf
import numpy as np 
```

```python
with open("data.txt") as f:
    content=f.read()
print(content)
```

```python
def preprocess(text):
    tokenizer = tf.keras.preprocessing.text.Tokenizer(char_level=True)
    tokenizer.fit_on_texts([text])

    max_id = len(tokenizer.word_index)

    encoded = np.array(
        tokenizer.texts_to_sequences([text])[0]
    ) - 1

    return encoded, max_id,tokenizer 
```

```python
encoded, max_id, tokenizer= preprocess(content)

seq_length = 100

x = []
y = []

for i in range(len(encoded) - seq_length):
    x.append(encoded[i:i+seq_length])
    y.append(encoded[i+1:i+seq_length+1])

x = np.array(x)
y = np.array(y)

print(x.shape)
print(y.shape)
```

```python
x = tf.one_hot(x, depth=max_id)
x
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.GRU(
        128,
        return_sequences=True,
        input_shape=[None, max_id],
        dropout=0.2,
        recurrent_dropout=0.2
    ),
    tf.keras.layers.GRU(
        128,
        return_sequences=True,
        dropout=0.2,
        recurrent_dropout=0.2
    ),
    tf.keras.layers.TimeDistributed(
        tf.keras.layers.Dense(max_id, activation="softmax")
    )
])
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam"
)
```

```python
model.fit(x,y,epochs=20)
```

```python
def next_char(text, temperature=1):
    x_new = np.array(
        tokenizer.texts_to_sequences([text])[0]
    ) - 1

    x_new = tf.one_hot([x_new], depth=max_id)

    y_proba = model.predict(x_new, verbose=0)

    y_proba = y_proba[:, -1, :]

    logits = tf.math.log(y_proba + 1e-10) / temperature

    char_id = tf.random.categorical(logits, 1)

    char_id = int(char_id.numpy()[0, 0])

    return tokenizer.index_word[char_id + 1]
```

```python
def complete_text(text, n_chars=50, temperature=1):
    for _ in range(n_chars):
        text += next_char(text, temperature)
    return text
```

```python
print(complete_text("f",temperature=0.2))
```

```python
print(complete_text("w",temperature=1))
```
